# Smart Document Assistant (RAG) with Llama 2

**Goal:** Build a system that can answer questions based on PDF documents using:
- **Retrieval-Augmented Generation (RAG)**
- **FAISS vector database**
- **Llama 2 local model**
- **PromptTemplate for concise answers**

**Pipeline:**
PDFs → Clean → Chunk → Embeddings → FAISS → Retriever → Llama 2 → Answer

Install dependencies outside the notebook with `python -m pip install -r requirements.txt`.

# Importing libraries

```python
import os
import re
from pathlib import Path

from langchain.chains import RetrievalQA
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.llms import CTransformers
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

# Functions for cleaning texts

```python
def clean_text(text: str) -> str:
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'\[\d+\]', '', text)
    text = re.sub(r'[^\w\s.,?!-]', '', text)
    return text.strip()

def is_useful_page(text: str, min_length: int = 50) -> bool:
    return len(text) > min_length
```

# Downloading and processing PDF files

```python
def resolve_path(value: str, default: Path, project_root: Path) -> Path:
    path = Path(value).expanduser()
    return path if path.is_absolute() else project_root / path


def positive_int_env(name: str, default: int) -> int:
    raw_value = os.getenv(name, str(default))
    try:
        value = int(raw_value)
    except ValueError as exc:
        raise ValueError(f"{name} must be a positive integer, got {raw_value!r}") from exc
    if value <= 0:
        raise ValueError(f"{name} must be a positive integer, got {value}")
    return value


PROJECT_ROOT = resolve_path(
    os.getenv("PROJECT_ROOT", str(Path.cwd())), Path.cwd(), Path.cwd()
).resolve()
DATA_PATH = resolve_path(
    os.getenv("DATA_DIR", str(PROJECT_ROOT / "data")), PROJECT_ROOT / "data", PROJECT_ROOT
)
MAX_FILES = positive_int_env("MAX_FILES", 270)

if not DATA_PATH.is_dir():
    raise FileNotFoundError(
        f"No data directory found at {DATA_PATH}. Download the PDFs and set DATA_DIR."
    )
pdf_paths = sorted(
    (path for path in DATA_PATH.rglob("*") if path.is_file() and path.suffix.lower() == ".pdf"),
    key=lambda path: path.relative_to(DATA_PATH).as_posix(),
)[:MAX_FILES]
if not pdf_paths:
    raise FileNotFoundError(
        f"No PDF files found under {DATA_PATH}. Download the dataset before ingestion."
    )

documents = []
for processed, pdf_path in enumerate(pdf_paths, 1):
    loader = PyPDFLoader(str(pdf_path))
    pages = loader.load()
    for doc in pages:
        cleaned = clean_text(doc.page_content)
        if is_useful_page(cleaned):
            doc.page_content = cleaned
            doc.metadata["source"] = str(pdf_path.relative_to(DATA_PATH))
            documents.append(doc)
    if processed % 10 == 0:
        print(f"{processed} PDFs processed")

if not documents:
    raise ValueError("The discovered PDFs contained no usable pages after cleaning.")

print(f"Total PDFs loaded: {len(pdf_paths)}")
print(f"Total pages loaded: {len(documents)}")
```

# Splitting texts into chunks

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)
print(f"Total text chunks: {len(chunks)}")
```

# Creating embeddings and FAISS vector store

```python
INDEX_PATH = resolve_path(
    os.getenv("FAISS_INDEX_DIR", str(PROJECT_ROOT / "faiss_index_arxiv")),
    PROJECT_ROOT / "faiss_index_arxiv",
    PROJECT_ROOT,
)
embeddings = HuggingFaceEmbeddings(
    model_name=os.getenv(
        "EMBEDDING_MODEL", "sentence-transformers/all-MiniLM-L6-v2"
    ),
    model_kwargs={"device": os.getenv("EMBEDDING_DEVICE", "cpu")},
)
vector_store = FAISS.from_documents(chunks, embeddings)
INDEX_PATH.mkdir(parents=True, exist_ok=True)
vector_store.save_local(str(INDEX_PATH))
print(f"FAISS vector store saved to {INDEX_PATH}")
```

# Preparing the Llama 2 model

```python
MODEL_DIR = resolve_path(
    os.getenv("LLM_MODEL_DIR", str(PROJECT_ROOT / "models" / "llama-2-7b-chat")),
    PROJECT_ROOT / "models" / "llama-2-7b-chat",
    PROJECT_ROOT,
)
MODEL_FILE = os.getenv("LLM_MODEL_FILE", "llama-2-7b-chat.Q4_K_M.gguf")
if Path(MODEL_FILE).is_absolute() or Path(MODEL_FILE).name != MODEL_FILE:
    raise ValueError("LLM_MODEL_FILE must be a filename inside LLM_MODEL_DIR")
MODEL_PATH = MODEL_DIR / MODEL_FILE
if not MODEL_PATH.is_file():
    raise FileNotFoundError(
        f"LLM model file not found at {MODEL_PATH}. Set LLM_MODEL_DIR/LLM_MODEL_FILE."
    )
llm = CTransformers(
    model=str(MODEL_DIR),
    model_file=MODEL_FILE,
    model_type=os.getenv("LLM_MODEL_TYPE", "llama"),
    config={
        "temperature": 0.0,
        "max_new_tokens": positive_int_env("LLM_MAX_NEW_TOKENS", 256),
        "context_length": positive_int_env("LLM_CONTEXT_LENGTH", 2048),
    },
)
```

# Setting up Prompts

```python
qa_prompt = PromptTemplate(
    input_variables=["context", "question"],
    template="""
You are a STRICT document-based QA system.

Rules:
1. Answer ONLY using information explicitly stated in the context.
2. Do NOT use external knowledge.
3. Do NOT infer or generalize.
4. If the answer is NOT mentioned verbatim, reply ONLY with:
   "Not mentioned in the documents."

Context:
{context}

Question:
{question}

Answer (bullet points only):
"""
)
```

# Creating a RAG Chain

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",   
    retriever=vector_store.as_retriever(
        search_kwargs={"k": 5}
    ),
    return_source_documents=True,
    chain_type_kwargs={
        "prompt": qa_prompt
    }
)

print("RAG system is ready")
```

# System Experience

```python
my_question = "What are the limitations of Large Language Models?"

response = qa_chain.invoke({"query": my_question})

print("\nAnswer:")
print(response["result"])

print("\nSources:")
for i, doc in enumerate(response["source_documents"], 1):
    print(f"{i}. {doc.metadata['source']} (page {doc.metadata.get('page', 'Unknown')})")
```

# Function to evaluate answers against documents

```python
def context_overlap_score(answer: str, docs: list) -> float:
    """Return an exact-token overlap diagnostic, not an answer-quality metric."""

    context_tokens = set(
        re.findall(r"\b\w+\b", " ".join(getattr(doc, "page_content", "") for doc in docs).lower())
    )
    answer_tokens = re.findall(r"\b\w+\b", answer.lower())
    if not answer_tokens:
        return 0.0
    return sum(token in context_tokens for token in answer_tokens) / len(answer_tokens)


score = context_overlap_score(
    response["result"],
    response["source_documents"]
)

print(f"\nAnswer–Context Token Overlap Diagnostic: {score:.2f}")
```
